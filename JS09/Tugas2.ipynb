```python
# Load data

import numpy as np
import pandas as pd

df = pd.read_csv('spam.csv', encoding='latin-1') # spesifiksi encoding diperlukan karena data tidak menggunakan UTF-8

df.head()
```

Buatlah model klasfikasi Multinomial Naive Bayes dengan ketentuan,

- Menggunakan data spam.csv

- Fitur CountVectorizer dengan mengaktifkan stop_words

- Evaluasi hasilnya

```python
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import accuracy_score, classification_report

df = df.drop(df.iloc[:, 2:], axis=1)
df.columns = ['Labels', 'SMS']

# Encode label (spam: 1, ham: 0)
new_labels = {'spam': 1, 'ham': 0}
df['Labels'] = df['Labels'].map(new_labels)

# Pisahkan fitur dan label
X = df['SMS'].values
y = df['Labels'].values

# Split data training dan testing (80:20)
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=50)

# Ekstraksi Fitur menggunakan CountVectorizer dengan stop_words
bow = CountVectorizer(stop_words='english')
X_train_bow = bow.fit_transform(X_train)
X_test_bow = bow.transform(X_test)

# Training dan Evaluasi Model MultinomialNB
mnb_bow = MultinomialNB()
mnb_bow.fit(X_train_bow, y_train)

# Prediksi dan evaluasi
y_pred_train_bow = mnb_bow.predict(X_train_bow)
y_pred_test_bow = mnb_bow.predict(X_test_bow)

acc_train_bow = accuracy_score(y_train, y_pred_train_bow)
acc_test_bow = accuracy_score(y_test, y_pred_test_bow)

print(f'Akurasi Data Train (CountVectorizer + Stop Words): {acc_train_bow:.4f}')
print(f'Akurasi Data Test (CountVectorizer + Stop Words): {acc_test_bow:.4f}')
```

Buatlah model klasfikasi Multinomial Naive Bayes dengan ketentuan,

- Menggunakan data spam.csv

- Fitur TF-IDF dengan mengaktifkan stop_words

- Evaluasi hasilnya dan bandingkan dengan hasil pada Tugas no 2.

- Berikan kesimpulan fitur mana yang terbaik pada kasus data spam.csv

```python
from sklearn.feature_extraction.text import TfidfVectorizer

# 1. Ekstraksi Fitur menggunakan TfidfVectorizer dengan stop_words
tfidf = TfidfVectorizer(stop_words='english')
X_train_tfidf = tfidf.fit_transform(X_train)
X_test_tfidf = tfidf.transform(X_test)

# 2. Training dan Evaluasi Model MultinomialNB
mnb_tfidf = MultinomialNB()
mnb_tfidf.fit(X_train_tfidf, y_train)

# Prediksi dan evaluasi
y_pred_train_tfidf = mnb_tfidf.predict(X_train_tfidf)
y_pred_test_tfidf = mnb_tfidf.predict(X_test_tfidf)

acc_train_tfidf = accuracy_score(y_train, y_pred_train_tfidf)
acc_test_tfidf = accuracy_score(y_test, y_pred_test_tfidf)

print(f'Akurasi Data Train (TF-IDF + Stop Words): {acc_train_tfidf:.4f}')
print(f'Akurasi Data Test (TF-IDF + Stop Words): {acc_test_tfidf:.4f}')
```

Analisis dan Kesimpulan:

- Performa Keseluruhan: CountVectorizer menghasilkan akurasi yang lebih tinggi (98.30%) dibandingkan TF-IDF (96.05%).

- Kemampuan Deteksi Spam (Recall): CountVectorizer jauh lebih baik dalam mendeteksi pesan spam secara keseluruhan (recall 0.90 berbanding 0.73 pada TF-IDF), yang berarti lebih sedikit pesan spam yang lolos (false negatives lebih sedikit).

- Fitur Terbaik: Berdasarkan hasil evaluasi di atas, fitur terbaik dan paling optimal untuk kasus klasifikasi teks SMS spam.csv menggunakan algoritma Multinomial Naive Bayes adalah CountVectorizer dengan mengaktifkan stop_words='english'.
