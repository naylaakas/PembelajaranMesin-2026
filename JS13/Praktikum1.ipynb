```python
import numpy as np

# Dataset XOR
X = np.array([[0,0],[0,1],[1,0],[1,1]])
y = np.array([[0],[1],[1],[0]])

# Parameter
input_size = 2
hidden_size = 2
output_size = 1
lr = 0.1

# Inisialisasi bobot
W1 = np.random.randn(input_size, hidden_size)
b1 = np.zeros((1, hidden_size))
W2 = np.random.randn(hidden_size, output_size)
b2 = np.zeros((1, output_size))

# Fungsi aktivasi
def sigmoid(x):
    return 1 / (1 + np.exp(-x))

def sigmoid_derivative(x):
    return x * (1 - x)

# Training
for epoch in range(10000):
    # Forward pass
    z1 = np.dot(X, W1) + b1
    a1 = sigmoid(z1)
    z2 = np.dot(a1, W2) + b2
    a2 = sigmoid(z2)

    # Hitung error
    error = y - a2

    # Backpropagation
    d_a2 = error * sigmoid_derivative(a2)
    d_W2 = np.dot(a1.T, d_a2)
    d_b2 = np.sum(d_a2, axis=0, keepdims=True)

    d_a1 = np.dot(d_a2, W2.T) * sigmoid_derivative(a1)
    d_W1 = np.dot(X.T, d_a1)
    d_b1 = np.sum(d_a1, axis=0, keepdims=True)

    # Update bobot
    W1 += lr * d_W1
    b1 += lr * d_b1
    W2 += lr * d_W2
    b2 += lr * d_b2

    if epoch % 1000 == 0:
        loss = np.mean(np.square(error))
        print(f"Epoch {epoch}, Loss: {loss}")

# Output akhir
print("Prediksi:")
print(a2)
```

- Ubah jumlah neuron hidden layer menjadi 3.

- Bandingkan hasil loss dengan konfigurasi awal.

- Tambahkan fungsi aktivasi ReLU dan bandingkan hasil.

```python
import numpy as np

# Dataset XOR
X = np.array([[0,0],[0,1],[1,0],[1,1]])
y = np.array([[0],[1],[1],[0]])

# Parameter umum
input_size = 2
output_size = 1
lr = 0.1
epochs = 10000

# Fungsi Aktivasi & Turunannya
def sigmoid(x):
    return 1 / (1 + np.exp(-x))

def sigmoid_derivative(x):
    return x * (1 - x)

def relu(x):
    return np.maximum(0, x)

def relu_derivative(x):
    return np.where(x > 0, 1, 0)

# Fungsi untuk training model
def train_nn(hidden_size, activation="sigmoid"):
    np.random.seed(42) # Seed agar inisialisasi bobot konsisten untuk perbandingan

    # Inisialisasi bobot dan bias
    W1 = np.random.randn(input_size, hidden_size) * 0.1
    b1 = np.zeros((1, hidden_size))
    W2 = np.random.randn(hidden_size, output_size) * 0.1
    b2 = np.zeros((1, output_size))

    for epoch in range(epochs):
        # 1. Forward Pass
        z1 = np.dot(X, W1) + b1
        if activation == "sigmoid":
            a1 = sigmoid(z1)
        elif activation == "relu":
            a1 = relu(z1)

        z2 = np.dot(a1, W2) + b2
        a2 = sigmoid(z2) # Output layer tetap menggunakan Sigmoid untuk biner [0, 1]

        # 2. Hitung Error
        error = y - a2

        # 3. Backpropagation
        d_a2 = error * sigmoid_derivative(a2)
        d_W2 = np.dot(a1.T, d_a2)
        d_b2 = np.sum(d_a2, axis=0, keepdims=True)

        if activation == "sigmoid":
            d_a1 = np.dot(d_a2, W2.T) * sigmoid_derivative(a1)
        elif activation == "relu":
            d_a1 = np.dot(d_a2, W2.T) * relu_derivative(a1)

        d_W1 = np.dot(X.T, d_a1)
        d_b1 = np.sum(d_a1, axis=0, keepdims=True)

        # 4. Update Bobot
        W1 += lr * d_W1
        b1 += lr * d_b1
        W2 += lr * d_W2
        b2 += lr * d_b2

    final_loss = np.mean(np.square(error))
    return final_loss, a2

# Run Eksperimen
loss_orig, pred_orig = train_nn(hidden_size=2, activation="sigmoid")
loss_h3_sig, pred_h3_sig = train_nn(hidden_size=3, activation="sigmoid")
loss_h3_relu, pred_h3_relu = train_nn(hidden_size=3, activation="relu")

print(f"=== HASIL EKSPERIMEN ===")
print(f"1. Hidden=2 (Sigmoid)  -> Loss Akhir: {loss_orig:.6f}")
print(f"2. Hidden=3 (Sigmoid)  -> Loss Akhir: {loss_h3_sig:.6f}")
print(f"3. Hidden=3 (ReLU)     -> Loss Akhir: {loss_h3_relu:.6f}\n")

print("Prediksi Output Akhir (Hidden=3, Sigmoid):")
print(np.round(pred_h3_sig, 4))
```
