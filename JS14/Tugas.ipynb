- Modifikasi model CNN pada praktikum 2 sehingga didapatkan akurasi testing lebih dari 80%.

- Buatlah model CNN untuk klasifikasi dataset MNIST.

```python
import tensorflow as tf
from tensorflow.keras import datasets, layers, models
from tensorflow.keras.preprocessing.image import ImageDataGenerator
import matplotlib.pyplot as plt

# 1. Unduh dan Preprocessing Dataset CIFAR-10
(train_images, train_labels), (test_images, test_labels) = datasets.cifar10.load_data()
train_images, test_images = train_images / 255.0, test_images / 255.0

# 2. Data Augmentation
datagen = ImageDataGenerator(
    width_shift_range=0.1,
    height_shift_range=0.1,
    horizontal_flip=True
)
datagen.fit(train_images)

# 3. Bangun Arsitektur CNN yang Ditingkatkan
model_mod = models.Sequential([
    # Block 1
    layers.Conv2D(32, (3, 3), padding='same', activation='relu', input_shape=(32, 32, 3)),
    layers.BatchNormalization(),
    layers.Conv2D(32, (3, 3), padding='same', activation='relu'),
    layers.BatchNormalization(),
    layers.MaxPooling2D((2, 2)),
    layers.Dropout(0.2),

    # Block 2
    layers.Conv2D(64, (3, 3), padding='same', activation='relu'),
    layers.BatchNormalization(),
    layers.Conv2D(64, (3, 3), padding='same', activation='relu'),
    layers.BatchNormalization(),
    layers.MaxPooling2D((2, 2)),
    layers.Dropout(0.3),

    # Block 3
    layers.Conv2D(128, (3, 3), padding='same', activation='relu'),
    layers.BatchNormalization(),
    layers.Conv2D(128, (3, 3), padding='same', activation='relu'),
    layers.BatchNormalization(),
    layers.MaxPooling2D((2, 2)),
    layers.Dropout(0.4),

    # Fully Connected
    layers.Flatten(),
    layers.Dense(128, activation='relu'),
    layers.BatchNormalization(),
    layers.Dropout(0.5),
    layers.Dense(10, activation='softmax')
])

# 4. Compile Model
model_mod.compile(
    optimizer='adam',
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)

# 5. Fit Model
epochs = 20
batch_size = 64
history_mod = model_mod.fit(
    datagen.flow(train_images, train_labels, batch_size=batch_size),
    epochs=epochs,
    validation_data=(test_images, test_labels)
)

# 6. Evaluasi Akurasi Testing
test_loss, test_acc = model_mod.evaluate(test_images, test_labels, verbose=2)
print(f"\n==========================================")
print(f"Akurasi Testing Hasil Modifikasi: {test_acc * 100:.2f}%")
print(f"==========================================")
```

```python
import tensorflow as tf
from tensorflow.keras import datasets, layers, models
import matplotlib.pyplot as plt

# 1. Unduh & Preprocessing Dataset MNIST
(train_images, train_labels), (test_images, test_labels) = datasets.mnist.load_data()

# Normalisasi piksel ke rentang 0-1 dan tambahkan channel dimension (28, 28, 1)
train_images = train_images.reshape((60000, 28, 28, 1)) / 255.0
test_images = test_images.reshape((10000, 28, 28, 1)) / 255.0

# 2. Bangun Model CNN untuk MNIST
model_mnist = models.Sequential([
    # Layer Konvolusi & Pooling 1
    layers.Conv2D(32, (3, 3), activation='relu', input_shape=(28, 28, 1)),
    layers.MaxPooling2D((2, 2)),

    # Layer Konvolusi & Pooling 2
    layers.Conv2D(64, (3, 3), activation='relu'),
    layers.MaxPooling2D((2, 2)),

    # Layer Konvolusi 3
    layers.Conv2D(64, (3, 3), activation='relu'),

    # Fully Connected Layer
    layers.Flatten(),
    layers.Dense(64, activation='relu'),
    layers.Dense(10, activation='softmax')
])

# 3. Cek Ringkasan Arsitektur Model
model_mnist.summary()

# 4. Compile Model
model_mnist.compile(
    optimizer='adam',
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)

# 5. Fit Model (5 Epoch sudah sangat cukup untuk MNIST)
history_mnist = model_mnist.fit(
    train_images, train_labels,
    epochs=5,
    validation_data=(test_images, test_labels)
)

# 6. Evaluasi Akurasi Model MNIST
test_loss_mnist, test_acc_mnist = model_mnist.evaluate(test_images, test_labels, verbose=2)
print(f"\n==========================================")
print(f"Akurasi Testing CNN pada MNIST: {test_acc_mnist * 100:.2f}%")
print(f"==========================================")
```
