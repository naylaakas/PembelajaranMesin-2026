```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
df_cafes = pd.read_csv('cafes.csv')
df_reviews = pd.read_csv('reviews.csv')
```

```python
# Menggabungkan kedua dataset menggunakan 'cafe_id' (Left Join)
df_merged = pd.merge(
    df_reviews,
    df_cafes,
    on='cafe_id',
    how='left',
    suffixes=('_review', '_cafe')
)

display(df_cafes.head())
display(df_reviews.head())
```

```python
# 1. Pengecekan Dimensi
print("Dimensi cafes.csv  :", df_cafes.shape)
print("Dimensi reviews.csv:", df_reviews.shape)
print("Dimensi gabungan   :", df_merged.shape)

# 2. Pengecekan Tipe Data & Missing Values
print("\n--- Info & Null cafes.csv ---")
df_cafes.info()
print(df_cafes.isnull().sum())

print("\n--- Info & Null reviews.csv ---")
df_reviews.info()
print(df_reviews.isnull().sum())

# 3. Ringkasan Nilai Statistik
print("\n--- Statistik cafes.csv ---")
display(df_cafes.describe())

print("\n--- Statistik reviews.csv ---")
display(df_reviews.describe())
```

```python
#Inspeksi Distribusi Data

num_cols = ['rating', 'review_count']
for col in num_cols:
    plt.figure(figsize=(6, 4))
    plt.hist(df_cafes[col], bins=20, edgecolor='black')
    plt.title(f'Distribusi {col} (Cafes)')
    plt.xlabel(col)
    plt.ylabel('Frekuensi')
    plt.show()
```

```python
#Inspeksi Outlier

fig, axes = plt.subplots(1, 2, figsize=(10, 4))

axes[0].boxplot(df_cafes['review_count'].dropna(), vert=True)
axes[0].set_title('Boxplot review_count (Cafes)')
axes[0].set_ylabel('review_count')

axes[1].boxplot(df_reviews['likes'].dropna(), vert=True)
axes[1].set_title('Boxplot likes (Reviews)')
axes[1].set_ylabel('likes')

plt.tight_layout()
plt.show()
```

```python
#Inspeksi Komposisi Variabel Kualitatif

fig, axes = plt.subplots(2, 2, figsize=(12, 8))
axes = axes.flatten()

# 1. Rating ulasan
df_reviews['rating'].value_counts().sort_index().plot(kind='bar', ax=axes[0])
axes[0].set_title('Distribusi Rating Ulasan (Reviews)')
axes[0].set_xlabel('Rating'); axes[0].set_ylabel('Jumlah')

# 2. Kategori kafe
df_cafes['categories'].value_counts().plot(kind='bar', ax=axes[1])
axes[1].set_title('Distribusi Kategori (Cafes)')
axes[1].set_xlabel('Categories'); axes[1].set_ylabel('Jumlah')
axes[1].tick_params(axis='x', rotation=25)

# 3. Kelengkapan Phone
df_cafes['phone'].notnull().map({True: 'Ada', False: 'Kosong (NaN)'}).value_counts().plot(kind='bar', ax=axes[2])
axes[2].set_title('Kelengkapan Nomor Telepon (Cafes)')
axes[2].set_ylabel('Jumlah')

# 4. Kelengkapan Website
df_cafes['website'].notnull().map({True: 'Ada', False: 'Kosong (NaN)'}).value_counts().plot(kind='bar', ax=axes[3])
axes[3].set_title('Kelengkapan Website (Cafes)')
axes[3].set_ylabel('Jumlah')

plt.tight_layout()
plt.show()
```

```python
#Inspeksi Korelasi

import seaborn as sns # Import seaborn for enhanced visualization

num_only = df_merged[['rating_review', 'likes', 'rating_cafe', 'review_count', 'latitude', 'longitude']]
corr = num_only.corr(numeric_only=True)

plt.figure(figsize=(8, 6))
sns.heatmap(corr, cmap='coolwarm', vmin=-1, vmax=1, annot=True, fmt=".2f")
plt.title('Heatmap Korelasi (Dataset Gabungan)')
plt.tight_layout()
plt.show()

# Urutan korelasi terhadap rating_review
print(corr['rating_review'].sort_values(ascending=False))

# Scatter Plot review_count vs rating (Cafes)
plt.figure(figsize=(7, 5))
plt.scatter(df_cafes['review_count'], df_cafes['rating'], alpha=0.7)
plt.xlabel('review_count')
plt.ylabel('rating')
plt.title('Scatter review_count vs rating (Cafes)')
plt.show()
```
