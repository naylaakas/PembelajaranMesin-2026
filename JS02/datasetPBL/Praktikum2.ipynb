```python
import pandas as pd
import numpy as np
```

```python
df_cafes = pd.read_csv('cafes.csv')
df_reviews = pd.read_csv('reviews.csv')

display(df_cafes.head())
```

```python
print("--- Missing Values cafes.csv ---")
print(df_cafes.isnull().sum())

print("\n--- Missing Values reviews.csv ---")
print(df_reviews.isnull().sum())
```

```python
# Imputasi kolom 'phone' dengan label "TIDAK ADA"
df_cafes['phone'] = df_cafes['phone'].fillna("TIDAK ADA")

# Imputasi kolom 'website' dengan label "TIDAK ADA"
df_cafes['website'] = df_cafes['website'].fillna("TIDAK ADA")

# Menggabungkan kembali dengan dataset reviews setelah bersih
df_merged = pd.merge(
    df_reviews,
    df_cafes,
    on='cafe_id',
    how='left',
    suffixes=('_review', '_cafe')
)
```

```python
# Validasi menggunakan info() dan isnull().sum()
df_cafes.info()
print("\nJumlah null setelah imputasi (cafes.csv):")
print(df_cafes.isnull().sum())

# Cek langsung sampel data yang telah diimputasi
display(df_cafes[['name', 'phone', 'website']].head(10))
```
