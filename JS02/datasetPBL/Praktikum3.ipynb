```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder, StandardScaler
```

```python
# Langkah 1 - Load Data (cukup 1 file hasil Praktikum 2)

dpath = 'Cafe-Reviews-fixed.csv'
df = pd.read_csv(dpath)
df.head()
```

```python
# Langkah 2 - Slice Data

df = df[['rating_review', 'rating_cafe', 'review_count', 'categories', 'website']]
df.head()
```

```python
# Langkah 3 - Encoding

le = LabelEncoder()
df['categories'] = le.fit_transform(df['categories'])
df['website'] = le.fit_transform(df['website'])
```

```python
# Langkah 4 - Verifikasi Hasil

df.head()
```

```python
# Langkah 5 - Standarisasi

std = StandardScaler()
df['review_count'] = std.fit_transform(df[['review_count']])
```
