```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder, StandardScaler
from sklearn.model_selection import train_test_split

# Load dataset gabungan yang sudah dibersihkan
df = pd.read_csv('Cafe-Reviews-fixed.csv')
df.head()
```

```python
kolom_tidak_dipakai = [
    'review_id', 'cafe_id', 'author_name', 'text', 'published_at',
    'scraped_at_review', 'name', 'address', 'phone', 'website',
    'scraped_at_cafe', 'source_query'
]

df = df.drop(columns=kolom_tidak_dipakai)
df.head()
```

```python
le = LabelEncoder()
df['categories'] = le.fit_transform(df['categories'])

# Cek hasil encoding
# 0: ['Grosir Kopi'], 1: ['Kafe'], 2: ['Kedai Kopi'], 3: ['Toko Pakaian'], 4: ['Toko']
df.head()
```

```python
num_cols = ['rating_review', 'likes', 'latitude', 'longitude', 'rating_cafe', 'review_count']

std = StandardScaler()
df[num_cols] = std.fit_transform(df[num_cols])

df.head()
```

```python
df_train, df_test = train_test_split(
    df,
    test_size=0.2,
    random_state=0,
    stratify=df['categories']
)

# Cek ukuran data
print(f'Jumlah data asli : {df.shape[0]}')
print(f'Jumlah data train: {df_train.shape[0]}')
print(f'Jumlah data test : {df_test.shape[0]}')

# Cek proporsi label pada masing-masing split
print('=========')
print(f'Jumlah label data asli:\n{df.categories.value_counts()}')
print(f'Jumlah label data train:\n{df_train.categories.value_counts()}')
print(f'Jumlah label data test:\n{df_test.categories.value_counts()}')
```
