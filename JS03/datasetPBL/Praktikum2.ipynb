```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder, StandardScaler

# Memuat dataset
df_cafes = pd.read_csv('cafes.csv')
df_reviews = pd.read_csv('reviews.csv')

# Menggabungkan kedua dataset
df = pd.merge(
    df_reviews,
    df_cafes,
    on='cafe_id',
    how='left',
    suffixes=('_review', '_cafe')
)

df.head()
```

```python
# --- 2. Encoding fitur 'categories' ---
cat_map = {
    "['Kedai Kopi']": 0,
    "['Kafe']": 1,
    "['Grosir Kopi']": 2,
    "['Toko']": 3,
    "['Toko Pakaian']": 4
}
df['categories'] = df['categories'].map(cat_map)

print("=== Hasil Setelah Encoding 'categories' ===")
print(df[['categories']].head(), "\n")
```

```python
# --- 3. Encoding fitur 'website' ---
df['website'] = df['website'].fillna('Unknown')
print("=== Hasil Setelah Fill NaN pada 'website' ===")
print(df[['website']].head(5), "\n")

df['website'] = df['website'].apply(lambda x: x.split('.')[-1] if x != 'Unknown' else 'U')
print("=== Hasil Ambil kode domain 'website' ===")
print(df[['website']].head(5), "\n")
```

```python
# --- 4. Imputasi fitur 'review_count' dengan median ---
df['review_count'] = df[['review_count']].fillna(df['review_count'].median())
print("=== Hasil Setelah Imputasi Data pada review_count dengan median ===")
print(df[['review_count']].head(10), "\n")
```

```python
scaler = StandardScaler()
df['review_count'] = scaler.fit_transform(df[['review_count']])
print("=== Hasil Setelah Standardisasi 'review_count' ===")
print(df[['review_count']].head(10), "\n")
```

```python
# Cek hasil seluruh kolom setelah pre-processing
print("=== Hasil Preprocessing (5 baris pertama) ===")
display(df.head())

# Langkah 9 - Seleksi Fitur (5 variabel)
df = df[['rating_review', 'rating_cafe', 'review_count', 'categories', 'website']]
df.head()
```
