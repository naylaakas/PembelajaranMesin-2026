Random Split

```python
import pandas as pd

df = pd.read_csv('Cafe-Reviews-fixed.csv')
df.head()
```

```python
# Split data
from sklearn.model_selection import train_test_split

# Split data training dan lainnya (rasio 8:1:1)
df_train, df_unseen = train_test_split(df, test_size=0.2, random_state=0)

# Split lagi antara validasi dan testing
df_val, df_test = train_test_split(df_unseen, test_size=0.5, random_state=0)

# Cek masing-masing ukuran data
print(f'Jumlah data asli: {df.shape[0]}')
print(f'Jumlah data train: {df_train.shape[0]}')
print(f'Jumlah data val: {df_val.shape[0]}')
print(f'Jumlah data test: {df_test.shape[0]}')

# Cek rasio tiap label
print('=========')
print(f'Jumlah label data asli:\n{df.rating_review.value_counts()}')
print(f'Jumlah label data train:\n{df_train.rating_review.value_counts()}')
print(f'Jumlah label data val:\n{df_val.rating_review.value_counts()}')
print(f'Jumlah label data test:\n{df_test.rating_review.value_counts()}')
```

Stratified Split

```python
import pandas as pd

df2 = pd.read_csv('Cafe-Reviews-fixed.csv')
df2.head()
```

```python
# Split data
from sklearn.model_selection import train_test_split

# Split data training dan lainnya dengan stratifikasi
df2_train, df2_unseen = train_test_split(df2, test_size=0.2, random_state=0, stratify=df2['rating_review'])

# Split lagi antara validasi dan testing dengan stratifikasi
df2_val, df2_test = train_test_split(df2_unseen, test_size=0.5, random_state=0, stratify=df2_unseen['rating_review'])

# Cek rasio tiap label
print(f'Jumlah label data asli:\n{df2.rating_review.value_counts()}')
print(f'Jumlah label data train:\n{df2_train.rating_review.value_counts()}')
print(f'Jumlah label data val:\n{df2_val.rating_review.value_counts()}')
print(f'Jumlah label data test:\n{df2_test.rating_review.value_counts()}')
```

Cross Validation 1

```python
import pandas as pd

df3 = pd.read_csv('Cafe-Reviews-fixed.csv')
df3.head()
```

```python
# Implementasi k-fold cross validation (random) dengan training dan testing saja
from sklearn.model_selection import KFold

# Inisiasi obyek kfold
kf = KFold(n_splits=4)
print(f'Jumlah fold: {kf.get_n_splits()}')
print(f'Obyek KFold: {kf}')

# Lakukan splitting dengan KFold
kf_split = kf.split(df3)
print(f'Jumlah data df: {df3.shape[0]}')

# Cek index data tiap fold
for train_index, test_index in kf_split:
    print(f'Index train: {train_index} | Index test: {test_index}')
```

Cross Validation 2

```python
import pandas as pd

df4 = pd.read_csv('Cafe-Reviews-fixed.csv')
df4.head()
```

```python
# Implementasi k-fold cross validation (random) dengan training, validation, dan testing data
from sklearn.model_selection import train_test_split, KFold

# Split dulu antara data training dan testing dengan rasio 8:2
df4_train, df4_test = train_test_split(df4, test_size=0.2, random_state=0)

# Inisiasi obyek kfold
kf2 = KFold(n_splits=4)
print(f'Jumlah fold: {kf2.get_n_splits()}')
print(f'Obyek KFold: {kf2}')

# Lakukan splitting dengan KFold untuk data df4_train
kf2_split = kf2.split(df4_train)
print(f'Jumlah data df_train: {df4_train.shape[0]}')

# Cek index data tiap fold
for train_index, test_index in kf2_split:
    print(f'Index train: {train_index} | Index test: {test_index}')
```
