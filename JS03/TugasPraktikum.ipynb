```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder, StandardScaler
```

```python
df = pd.read_csv("wbc.csv")        #load dataset
```

```python
#Pisahkan antara variabel yang dapat digunakan dan variabel yang tidak dapat digunakan.

df_selected = df.drop(columns=["id", "Unnamed: 32"], errors="ignore")
```

```python
#Lakukan proses encoding pada kolom "diagnosis".

le = LabelEncoder()
df_selected["diagnosis"] = le.fit_transform(df_selected["diagnosis"])
```

```python
#Lakukan proses standarisasi pada semua kolom yang memiliki nilai numerik.

feature_cols = df_selected.columns.drop("diagnosis")
scaler = StandardScaler()
df_selected[feature_cols] = scaler.fit_transform(df_selected[feature_cols])
```

```python
#Lakukan proses stratified split data untuk membuat data latih dan data uji dengan rasio 80:20.

X = df_selected.drop(columns=["diagnosis"])      #pisahkan fitur dan target
y = df_selected["diagnosis"]
```

```python
X_train, X_test, y_train, y_test = train_test_split(      #bagi data untuk data latih (80%) dan data uji (20%)
    X, y, test_size=0.2, random_state=42, stratify=y)
```

```python
summary_df = pd.DataFrame(              #tabel pembagian data
    {
        "Dataset": ["Data Latih (Train)", "Data Uji (Test)", "Total"],
        "Jumlah Baris": [len(X_train), len(X_test), len(df)],
        "Persentase": ["80%", "20%", "100%"],
        "Malignant (1)": [sum(y_train == 1), sum(y_test == 1), sum(y == 1)],
        "Benign (0)": [sum(y_train == 0), sum(y_test == 0), sum(y == 0)],
    }
)

display(summary_df)
```
